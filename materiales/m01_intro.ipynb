### 1. ¿Qué significa programar?

- Programar consiste, en términos muy sencillos, en **darle instrucciones a una computadora para realizar una tarea**. 

- Estas instrucciones se escriben utilizando un lenguaje de programación, que nos permite expresar de manera estructurada qué queremos que haga la computadora.

- En análisis de datos podemos pensar en este proceso a partir de tres elementos:

    - **Datos → Instrucciones → Resultados**
    - Por ejemplo, tenemos una base de datos con información sobre la edad de la población de distintos municipios y queremos saber ¿cuál es la edad promedio?
        - **Datos:** la edad de la población de cada municipio.
        - **Instrucciones:** indicarle a la computadora qué datos utilizar y que calcule su promedio.
        - **Resultado:** el valor de la edad promedio.

- Una de las ventajas de programar es que podemos construir una secuencia de instrucciones para realizar tareas más complejas

- **Repetir el mismo análisis con nuevos datos sin tener que realizar nuevamente cada operación de manera manual**.

### 1.3 ¿Dónde ejecutamos Python?

Python puede ejecutarse de diferentes maneras pero comenzaremos con notebooks dentro de **Google Colab**, y posteriormente aprenderemos a ejecutar Python directamente desde nuestras computadoras.

#### Notebooks

Un notebook es un tipo de archivo interactivo que nos permite escribir y ejecutar código en Python (y otros lenguajes) de manera organizada. Está compuesto por celdas, que pueden ser de distintos tipos:

- **Celdas de texto**: permiten escribir explicaciones, interpretar tablas o gráficas, documentar resultados y mantener una estructura clara y entendible para compartir con otras personas. Se escriben en Markdown, lo que permite dar formato (negritas, itálicas, listas, títulos, etc.).

- **Celdas de código**: son las que se utilizan para escribir y ejecutar instrucciones en Python. Los resultados aparecen justo debajo de la celda.

- Además de los notebooks, también existen los archivos .py, que contienen
únicamente código ejecutable.

### 1.4 Celdas de ejecución

Una característica importante de los notebooks es que **cada celda de código puede ejecutarse de manera independiente**.

Por ejemplo, podemos tener una primera celda:

```python
nombre = "Ana"
```

```python
print(nombre)
```

La segunda celda funciona porque anteriormente ejecutamos la celda donde definimos `nombre`.

Esto significa que **el orden en el que ejecutamos las celdas importa**. Un notebook puede verse ordenado de arriba hacia abajo, pero Python recuerda las instrucciones en el orden en que realmente fueron ejecutadas.

Por esta razón, una buena práctica será procurar que nuestros notebooks puedan ejecutarse **de principio a fin y en orden**.

```python
print(hola)
```

```python
hola = 1
```
