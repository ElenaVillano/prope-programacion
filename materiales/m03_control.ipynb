### 1. Condicionales

Los condicionales nos ayudan a realizar tareas basadas en una condiciones, se utiliza `if`, `elif`, `else`. Es decir, nuestro programa toma decisiones dependiendo de si una condición es `True` o `False`. 

- `if` es el que se usa para checar si una sola condición sucede y se ejecuta si es `True`, si es `False` no se ejecuta. 

````python
temperatura = 35

#Intenta 20 y 32 de temperatura.
###
if temperatura >= 30:
    print('\n Hace calor \U0001F31E ¡Vamos a la alberca! :D')
````

````python
temperatura = 35

if temperatura >= 30:
    print('\n Hace calor')
    print(' \U0001F31E')
    print(' ¡Vamos a la alberca! :D')

print("\n Fin del programa")
````

### 2. Loops

Los loops se utilizan para hacer ciertas tareas múltiples veces, y tienen la habilidad de iterar sobre los elementos de cualquier secuencia como listas, tuplas, strings o las llaves de los diccionarios. 

**Repiten una misma instrucción varias veces**

En lugar de escribir:

````python
print("Monterrey")
print("Guadalajara")
print("Puebla")
print("Mérida")
print("Oaxaca")
````

podemos utilizar un `for`

**Para cada valor en la secuencia, realiza esta tarea.**

```
for variable in secuecia:
    tareas(s)
```

````python
municipios = ["Monterrey", "Guadalajara", "Puebla", "Mérida", "Oaxaca"]

for municipio in municipios:
    print(municipio)
````

````python
colores = ['Rojo', 'Negro', 'Azul', 'Blanco', 'Rosa']
for color in colores:
    print(color)
````

````python
edades = [17, 25, 42, 68]

for edad in edades:
    print(edad + 1)
````

````python
# Combinar loops con condicionales
edades = [17, 25, 42, 68]

for edad in edades:
    if edad >= 18:
        print(edad, "Mayor de edad")
    else:
        print(edad, "Menor de edad")
````

````python
# Construir resultados con un loop para análisis de datos

edades = [17, 25, 42, 68]
edades_futuras = []

for edad in edades:
    edades_futuras.append(edad + 10)

edades_futuras
````

#### `range()`

Genera una secuencia de números que podemos utilizar en un loop.


````python
for numero in range(5):
    print(numero)
````

````python
# Iniciar desde 1

for numero in range(1, 6):
    print(numero)
````

````python
# Especificar tamaño del paso
for numero in range(0, 11, 2):
    print(numero)
````

#### `enumerate()`

Es útil cuando queremos conocer **tanto el elemento como su posición.**

````python
lista = [3,5,7,9,11]

# Si necesitas el valor e índice
for index, value in enumerate(lista):
    print("La posición", index, "en la lista es ", value)
````

#### `while`

Repite instrucciones mientras una condición sea verdadera

````python
numero = 1

while numero <= 5:
    print(numero)
    numero = numero + 1
````

### 3. Funciones

Una función nos permite **agrupar una serie de instrucciones para poder reutilizarlas**.

Las funciones hacen que tu código sea más eficiente, funcional y se pueda leer fácilmente, sobre todo hacen que tu código sea reusable. 

Supongamos que constantemente necesitamos determinar si alguien es mayor de edad:

````python
edad = 25

if edad >= 18:
    print("Mayor de edad")
else:
    print("Menor de edad")
````

Pero si necesitamos hacerlo muchas veces, podemos crear una función:

````python
def clasificar_edad(edad):
    if edad >= 18:
        print("Mayor de edad")
    else:
        print("Menor de edad")
````

````python
clasificar_edad(25)
````

````python
clasificar_edad(15)
````

Una función tiene las siguientes partes:
* palabra clave `def` seguida del nombre de la función;
* argumentos o parámetros (puede no haber);
* la tarea que hará la función;
* opcionalmente, puede devolver un objeto(s) a través del la palabra clave `return`.

Primero declaras la función con los puntos anteriores, después puedes llamar a la función con el nombre de tu función y los elementos necesarios para correrla. 

```
def nombre_funcion(parametros):
    tareas(s)
```

````python
# Declaras la función
def saludo():
    print("Hooliii! Bienvenido a la fiesta de python.")
````

````python
# Llamas la función
saludo()

# Ojo que aquí no hay ningún parámetro
````

`return`

Muchas veces no queremos solamente imprimir un resultado. Queremos que la función nos devuelva un valor que podamos seguir utilizando.

````python
def calcular_total(precio, cantidad):
    total = precio * cantidad
    return total
````

````python
calcular_total(20, 5)
````

````python
compra = calcular_total(30, 5)
````
