### 1. Reconocer diferentes formas de trabajar con Python

En Python vamos a encontrarnos constantemente con expresiones que se parecen, pero significan cosas diferentes.

#### Variables

Una variable contiene uno o varios valores

```python
edades = [15, 22, 67, 34]
edades
```

```python
edades*2
```

#### Funciones

Una función realiza una operación.

- `len` es la función
- `edades` es el argumento

```python
len(edades)
```

```python
type(edades)
```

#### Métodos

Los objetos de Python también pueden tener sus propias funciones, llamadas **métodos**.

**Importante**: 

- algunos métodos modifican directamente el objeto sobre el que trabajan, mientras que otros devuelven un nuevo resultado.
- Esto depende tanto del tipo de objeto como del método utilizado (revisar documentación de los métodos).
- Por ejemplo, las listas son **mutables** y `append()` modifica la lista original; los strings son **inmutables**, por lo que `upper()` devuelve un nuevo string.

```python
municipios = ["Monterrey", "Puebla"]

municipios.append("Oaxaca")
```

```python
municipios
```

```python
nombre = "elena"

nombre.upper()
```

```python
nombre
```

```python
# si queremos asignarlo
nombre = nombre.upper()

nombre
```

```python
# otro ejemplo

numeros = [3, 1, 2]

numeros.sort()

numeros
```

```python
numeros_1 = [3, 1, 2]

sorted(numeros_1)
```

```python
numeros_1
```

#### Atributos

Un atributo contiene información sobre un objeto y normalmente no lleva paréntesis.

Esto será mucho más evidente cuando trabajemos con paquetes `pandas` y `NumPy`. 

| Expresión    | ¿Qué es?        | Ejemplo         |
| ------------ | --------------- | --------------- |
| `x`          | Variable/objeto | `edades`        |
| `funcion(x)` | Función         | `len(edades)`   |
| `x.metodo()` | Método          | `edades.sort()` |
| `x.atributo` | Atributo        | `edades.shape`  |

## 3. Arreglos

En las sesiones anteriores hablamos de estructuras de almacenamiento de valores como listas, tuplas, etc. En este caso hablaremos de **arreglos** que son un tipo de estructura de dato que requiere **numpy (np)**. NumPy es un paquete de Python diseñado para trabajar de manera eficiente con datos numéricos.


- Los arregos son homogéneos, en el sentido de que solo pueden almacenar elementos de un solo tipo de dato (ya sea float, int o str). 
- Los arreglos tienden a ser más eficiencies para cómputo numérico y grandes cantidades de datos por la manera en la que se almacenan. 
- Permiten realizar operaciones vectoriales por lo que puedes hacer operaciones matemáticas sobre el arreglo completo. 
- Sus funciones y operaciones están especialmente optimizadas para cómputo numérico.



```python
# mandamos a llamar numpy
import numpy as np
```

```python
# Crear un arreglo
mi_arreglo = np.array([1, 2, 3, 4, 5])
print("Arreglo original:", mi_arreglo)
```

```python
type(mi_arreglo)
```

```python
# Realizamos operaciones con el arreglo que creamos
arreglo_cuadrado = mi_arreglo ** 2
print("Los valores del arreglo al cuadrado:", arreglo_cuadrado)
```

```python
suma_arreglo = np.sum(mi_arreglo)
print("Suma del arreglo:", suma_arreglo)
```

```python
media_arreglo = np.mean(mi_arreglo)
print("Media del arreglo:", media_arreglo)
```

```python
a_1 = np.array([8, 2, 9, 25, 4])
a_2 = np.array([66, 9, 17, 30, 5])
```

```python
result = a_2 - a_1
result
```
