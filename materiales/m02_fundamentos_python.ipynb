#### Operadores lógicos:

Los operadores lógicos nos permiten **combinar condiciones**.

Los tres principales son:

- `and`: ambas condiciones deben cumplirse.
- `or`: al menos una condición debe cumplirse.
- `not`: invierte el resultado de una condición.

##### `and`

````python
edad = 25
es_estudiante = True

edad >= 18 and es_estudiante == True
````

Podemos modificar alguno de los valores y observar qué sucede:

````python
edad = 16
es_estudiante = True

edad >= 18 and es_estudiante == True
````

Una forma más sencilla de escribir la misma condición es:

````python
edad >= 18 and es_estudiante 

# ¿Qué tengo que hacer para que el resultado sea True?
````

##### `or`

Con `or`, basta con que una de las condiciones sea verdadera.

````python
edad = 70
es_estudiante = False

edad >= 65 or es_estudiante
````

##### `not`

`not` invierte un valor booleano:

````python
es_estudiante = True

not es_estudiante
````

También puede utilizarse sobre una comparación:

````python
edad = 25

not edad < 18
````

#### Indexación

Cada elemento de una lista tiene una posición o índice.

Una característica importante de Python es que la indexación comienza en 0, no en 1.

````python
estados = ["Nuevo León", "Jalisco", "Puebla", "Oaxaca"]
#           0              1          2         3
````

````python
# Para obtener el primer elemento:
estados[0]
````

````python
# El segundo
estados[1]
````

````python
# El cuarto
estados[3]
````

También podemos contar desde el final utilizando número negativos:

````python
# Último elemento
estados[-1]
````

````python
# Penúltimo elemento
estados[-2]
````

#### Slicing

También podemos seleccionar varios elementos de una lista

````python
edades = [25, 32, 41, 19, 28, 35]

edades[1:4]

# Empieza en la posición 1 e incluye elementos hasta antes de la posición 4.
````

````python
# Los primeros tres elementos
edades[:3]
````

````python
# Desde la posición 3 hasta el final
edades[3:]
````

````python
# Toda la lista
edades[:]
````

#### Tuplas

Una tupla es similar a la lista en el sentido de que son valores ordenados, pero la diferencia es que las tuplas son **inmutables**, por lo que NO puedes cambiar sus elementos.  

Dependiento del contexto, las tuplas pueden ser de gran ayuda en lugar de listas. Primero, porque son más rápidas operacionalmente que las listas, y segundo porque son más seguras pues no se puede modificar una vez declarada. 

Éstas se definen con `( )`.

````python
mi_tupla = (99, 'Andrea', 7.6, 99)
````

````python
type(mi_tupla)
````

````python
mi_tupla
````

````python
# Segundo elemento de la tupla
mi_tupla[1]
````

````python
# número de elementos en tupla
len(mi_tupla)
````

#### Conjunto (set)

Es una secuencia de objetivos no ordenados, es menos famoso pero muy útil para diversas cosas. No tiene ningún valor repetido, a diferencia de las listas y las tuplas. En los conjuntos puedes realizar operaciones como unión, intersección. Para definir un conjunto utilizamos `{ }`

````python
mi_set = {9,10,11,11}
````

````python
mi_set
````

````python
type(mi_set)
````

````python
grupo1 = {"Alicia", "Pedro", "Juan", "Diana", "Eva"}
grupo2 = {"Pedro", "Diana", "Luis", "Ana", "Alejandra"}
grupo3 = {"Juan", "Eva", "Luis", "Ana", "Isaac"}
````

````python
# Intersección: 
miembros_comunes = grupo1.intersection(grupo2, grupo3)
print("Estudiantes que están los tres grupos:", miembros_comunes)
````

````python
# Intersección: 
miembros_comunes = grupo1.intersection(grupo2)
print("Estudiantes que están en grupo1 y grupo2:", miembros_comunes)
````

````python
# Unión: 
todos = grupo1.union(grupo2, grupo3)
print("Estudiantes únicos en todos los grupos:", todos)
````

````python
# Diferencia: 
exclusivos_g1 = grupo1.difference(grupo2, grupo3)
print("Estudiantes que están en el grupo1 pero no en 2 y 3:", exclusivos_g1)
````

####  Diccionario

Es una colección NO ordenada de pares de valores-llaves. Los diccionarios se utilizan principalmente para almacenar informaicón. Para definirlos utilizamos `{key: value}` o `llave → valor`

````python
persona = {
    "nombre": "Ana",
    "edad": 32,
    "estado": "Oaxaca",
    "es_estudiante": False
}

persona
````

En lugar de acceder a la información mediante una posición, utilizamos la llave:

````python
persona["nombre"]
````

````python
persona["edad"]
````

````python
# Podemos modificar un valor
persona["edad"] = 33

persona
````

````python
# Agregar nueva información
persona["ocupacion"] = "Investigadora"

persona
````

````python
# Consultar llaves
persona.keys()
````

````python
# valores
persona.values()
````

````python
# llaves y valores
persona.items()
````
