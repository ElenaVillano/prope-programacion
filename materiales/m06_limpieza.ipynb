### Seleccionando subconjuntos de la base

#### `.loc` 

Permite seleccionar filas y columnas al mismo tiempo.

La estructura general es:

```
df.loc[filas, columnas]
```



````python
df.loc[df["state"] == "Vermont",["state", "indicator", "value"]]
````

````python
df.loc[df["value"] > 40, ["state", "indicator","subgroup", "value"]]
````

#### `.iloc` 

Mientras `loc` trabaja principalmente con nombres y condiciones, `iloc` selecciona por posición.

Primeras 5 filas y primeras 4 columnas:

````python
df.iloc[:5, :4]
````

````python
# Selecciona las filas 2 hasta la 6
df.iloc[2:6,]
````

````python
df.iloc[:,[2,4,6]]
````

Para análisis de datos normalmente utilizaremos más `.loc`, porque los nombres de las variables hacen el código más fácil de leer.

### Ordenar datos

Podemos ordenar las observaciones con `sort_values()`.

````python
# Orden ascendente
df.sort_values("value")
````

````python
# Orden descendente
df.sort_values("value", ascending=False)
````

### Crear y modificar variables

````python
df["nueva_variable"] = ...
````

Para trabajar sobre una nueva

````python
df_ejemplo = df.copy()
````

````python
df_ejemplo
````

````python
df_ejemplo["amplitud_ic"] = (df_ejemplo["high_ci"] - df_ejemplo["low_ci"])

df_ejemplo[["low_ci", "high_ci", "amplitud_ic"]].sample(10)
````

````python
# Todas las comparaciones que realices se pueden convertir en variables
df_ejemplo["valor_mayor_50"] = (
    df_ejemplo["value"] > 50
)

df_ejemplo[
    ["value", "valor_mayor_50"]
].head()
````

````python
# Podemos crear una variable que indique si pertenece a ciertos estados:
df_ejemplo["estado_seleccionado"] = (
    df_ejemplo["state"]
    .isin(["Wyoming", "North Dakota", "Hawaii"])
)
````

````python
df_ejemplo.head()
````
